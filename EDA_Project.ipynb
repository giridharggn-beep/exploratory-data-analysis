**Exploratory Data Analysis on Sample Superstore Datasett**

## ***Introduction***

This project performs Exploratory Data Analysis (EDA) on the Sample Superstore dataset. The analysis focuses on sales, profit, discounts, categories, regions, and sub-categories to identify useful patterns and insights.

 ## Dataset Overview

The Sample Superstore dataset contains sales records with information about customers, products, regions, sales, quantity, discounts, and profit.

The dataset contains 9,994 original records and 13 columns. After removing 17 duplicate records, 9,977 records were used for the analysis.

```python
import pandas as pd
import matplotlib.pyplot as plt

print("EDA Project is ready")
```

```python
df = pd.read_csv("SampleSuperstore1.csv")

df.head()
```

## Exploratory Data Analysis

```python
df['Category'].value_counts()
```

```python
category_counts = df['Category'].value_counts()

category_counts.plot(kind='bar')

plt.title('Number of Records by Category')
plt.xlabel('Category')
plt.ylabel('Number of Records')
plt.xticks(rotation=0)

plt.show()
```

```python
sub_category_counts = df['Sub-Category'].value_counts()

sub_category_counts.plot(kind='bar')

plt.title('Number of Records by Sub-Category')
plt.xlabel('Sub-Category')
plt.ylabel('Number of Records')
plt.xticks(rotation=90)

plt.show()
```

```python
sales_by_category = df.groupby('Category')['Sales'].sum()

print(sales_by_category)
```

```python
sales_by_category.plot(kind='bar')

plt.title('Total Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)

plt.show()

```

```python
quantity_by_category = df.groupby('Category')['Quantity'].sum()

print(quantity_by_category)
```

```python
profit_by_category = df.groupby('Category')['Profit'].sum()

print(profit_by_category)
```

```python
profit_by_category.plot(kind='bar')

plt.title('Total Profit by Category')
plt.xlabel('Category')
plt.ylabel('Total Profit')
plt.xticks(rotation=0)

plt.show()
```

```python
region_counts = df['Region'].value_counts()

print(region_counts)
```

```python
region_counts.plot(kind='bar')

plt.title('Number of Records by Region')
plt.xlabel('Region')
plt.ylabel('Number of Records')
plt.xticks(rotation=0)

plt.show()
```

```python
sales_by_region = df.groupby('Region')['Sales'].sum()
print(sales_by_region)
```

```python
sales_by_region.plot(kind='bar')
plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.show()
```

```python
profit_by_region = df.groupby('Region')['Profit'].sum()
print(profit_by_region)
```

```python
profit_by_region.plot(kind='bar')
plt.title('Total Profit by Region')
plt.xlabel('Region')
plt.ylabel('Total Profit')
plt.xticks(rotation=0)
plt.show()
```

```python
avg_profit_category = df.groupby('Category')['Profit'].mean()
print(avg_profit_category)
```

```python
avg_profit_category.plot(kind='bar')
plt.title('Average Profit by Category')
plt.xlabel('Category')
plt.ylabel('Average Profit')
plt.xticks(rotation=0)
plt.show()
```

```python
plt.scatter(df['Discount'], df['Profit'], alpha=0.4)
plt.title('Discount vs Profit')
plt.xlabel('Discount')
plt.ylabel('Profit')
plt.show()
```

```python
top_subcategories = df.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=False).head(10)
print(top_subcategories)
```

```python
top_subcategories.sort_values().plot(kind='barh')

plt.title('Top 10 Sub-Categories by Sales')
plt.xlabel('Total Sales')
plt.ylabel('Sub-Category')
plt.show()
```

```python
top_profit_subcategories = df.groupby('Sub-Category')['Profit'].sum().sort_values(ascending=False).head(10)
print(top_profit_subcategories)
```

```python
top_profit_subcategories.sort_values().plot(kind='barh')

plt.title('Top 10 Sub-Categories by Profit')
plt.xlabel('Total Profit')
plt.ylabel('Sub-Category')
plt.show()
```

```python
profit_by_subcategory = df.groupby('Sub-Category')['Profit'].sum()

loss_making = profit_by_subcategory[profit_by_subcategory < 0]

print(loss_making)
```

```python
loss_making.sort_values().plot(kind='barh')

plt.title('Loss-Making Sub-Categories')
plt.xlabel('Total Profit')
plt.ylabel('Sub-Category')
plt.show()
```

```python
plt.scatter(df['Sales'], df['Profit'], alpha=0.4)

plt.title('Sales vs Profit')
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.show()
```

```python
avg_discount_category = df.groupby('Category')['Discount'].mean()
print(avg_discount_category)
```

```python
avg_discount_category.plot(kind='bar')

plt.title('Average Discount by Category')
plt.xlabel('Category')
plt.ylabel('Average Discount')
plt.xticks(rotation=0)

plt.show()
```

```python
category_summary = df.groupby('Category').agg(
    Sales=('Sales', 'sum'),
    Profit=('Profit', 'sum')
)

category_summary['Profit Margin (%)'] = (
    category_summary['Profit'] / category_summary['Sales']
) * 100

print(category_summary)
```

```python
category_summary['Profit Margin (%)'].plot(kind='bar')

plt.title('Profit Margin by Category')
plt.xlabel('Category')
plt.ylabel('Profit Margin (%)')
plt.xticks(rotation=0)

plt.show()
```

```python
top_subcategories = df.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=False).head(10)

print(top_subcategories)
```

```python
top_subcategories.sort_values().plot(kind='barh')

plt.title('Top 10 Sub-Categories by Sales')
plt.xlabel('Total Sales')
plt.ylabel('Sub-Category')

plt.show()
```

```python
top_profit_subcategories = df.groupby('Sub-Category')['Profit'].sum().sort_values(ascending=False).head(10)

print(top_profit_subcategories)
```

```python
top_profit_subcategories.sort_values().plot(kind='barh')

plt.title('Top 10 Sub-Categories by Profit')
plt.xlabel('Total Profit')
plt.ylabel('Sub-Category')

plt.show()
```

```python
profit_by_subcategory = df.groupby('Sub-Category')['Profit'].sum()

loss_making = profit_by_subcategory[profit_by_subcategory < 0]

print(loss_making)
```

```python
loss_making.sort_values().plot(kind='bar')

plt.title('Loss-Making Sub-Categories')
plt.xlabel('Total Profit')
plt.ylabel('Sub-Category')

plt.show()
```
